```python
print (f'Median value of horsepower column {df['horsepower'].median()}') # checking for median value in horse power column
print (f'Mode of horsepower column {df['horsepower'].mode()}') # checking for mode in horse power column

print ('handling null values for horsepower column')
df['horsepower'] = df['horsepower'].fillna(df['horsepower'].mode()[0])  # filling null values with mode
print(df['horsepower'].isna().sum())

print (f'Median value after missing value removal of horsepower column {df['horsepower'].median()}') 
```
